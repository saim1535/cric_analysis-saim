```python
df["Country"]=df["Country"].str.split(pat=")").str[0]
```

```python
df["High_Score"]=df["High_Score"].str.split(pat="*").str[0]
```

```python
df["Ball_face"]=df["Ball_face"].str.split(pat="+").str[0]
```

```python
df[df["Strike_Rate"]=="-"]
```

```python
df["0"]=df["0"].replace("-",0)
```

```python
df["Strike_Rate"]=df["Strike_Rate"].replace("-",0)
```

```python
df["Ball_face"]=df["Ball_face"].replace("-",0)
```

```python
df["High_Score"]=df["High_Score"].astype(int)
```

```python
df["Ball_face"]=df["Ball_face"].astype(int)
```

```python
df["Strike_Rate"]=df["Strike_Rate"].astype(float)
```

```python
df["0"]=df["0"].astype(int)
```

```python
df.info()
```

```python
df["starting_year"]=df["Span"].str.split(pat="-").str[0].astype(int)
```

```python
df["ending_year"]=df["Span"].str.split(pat="-").str[1].astype(int)
```

```python
df.head()
```

```python
df["Duration"]=df["ending_year"]-df["starting_year"]
```
